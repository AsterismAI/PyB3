```python
# use the astrostandards to parse data
lines = B3s.split('\n')
LU.ObsDll.ObsRemoveAll()
obids = [ LU.ObsDll.ObsAddFrB3Card( LU.Cstr(X,512) ) for X in lines ]

def getB3data( obid ):
    LU.ObsDll.ObsDataToArray( obid, XA_OBS.data )
    return XA_OBS.toDict()

astd_data = [ getB3data(X) for X in obids ]
# astd_data
```
